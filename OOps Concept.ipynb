```python
class Bank:
    ifsc_code = 1009
    def _init_(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance:{self.balance}")
        else:
            print("Deposit amount must be positive.")
            
    def withdraw(self, amount):
        if amount <=0:
            print("withdrawl amount must be positive.")
        elif amount > self.balance:
            print("Insufficient funds")
        else: 
            self.balance -= amount
            print(f"withdraw {amount}. remaining baalnce: {self.balance}")

    def display_balance(self):
        print(f"account holder{ self.owner}, Balance:{self.balance}, IFSC CODE : {self.ifsc_code}")
        
            
            

 
```
