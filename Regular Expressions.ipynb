```python
import re

count =0

pattern = re.compile("ab")
matcher = pattern.finditer("abaabbaaba")
for match in matcher:
    count = count +1
    print(match.start(), '...', match.end(), '...', match.group())

print('The number of occurences ' , count)
```

## character classes

```python
import re 
print('.................................')
x = "[^a-zA-Z0-9]"
print('Character classes : ', x)

matcher = re.finditer(x, "a7x@k9zZ")
for match in matcher :
    print(match.start(), '--------', match.group())
```

## pre-define Characer class

```python
import re 
print('.................................')
x = r"\W"
print('Character classes : ', x)

matcher = re.finditer(x, "a7x@k9zZ")
for match in matcher :
    print(match.start(), '--------', match.group())
```
