```python
import random

random_ages = [random.randint(5, 99) for _ in range(100)]
print("Random ages:" , random_ages)
```

```python
random_ages = [random.randint(5, 99) for _ in range(100)]
age_bins = [ 5, 20, 40,50, 60, 80]
plt.hist(random_ages, bins=age_bins, edgecolor='black' , color ='skyblue')
plt.xlabel('Age Groups')
plt.ylabel('Number of People')
plt.title('Histogram of Age Groups')
plt.xticks(age_bins)
plt.show()
```
